```python
from pathlib import Path
from torchvision.datasets import EuroSAT

dataset = EuroSAT(
    root=Path("../data"),
    download=True,
)

print("Immagini:", len(dataset))
print("Classi:", dataset.classes)

image, label = dataset[0]
print("Dimensioni:", image.size)
print("Classe prima immagine:", dataset.classes[label])
```

```python
from pathlib import Path
import matplotlib.pyplot as plt

data_dir = Path("../data")

print("Numero immagini:", len(dataset))
print("Classi:", dataset.classes)
print("Mappa classi:", dataset.class_to_idx)

image, label = dataset[0]

print("Tipo immagine:", type(image))
print("Dimensioni:", image.size)
print("Etichetta:", label)
print("Classe:", dataset.classes[label])
```

```python
fig, axes = plt.subplots(2, 5, figsize=(15, 6))

for ax, index in zip(axes.ravel(), range(10)):
    image, label = dataset[index]
    ax.imshow(image)
    ax.set_title(dataset.classes[label])
    ax.axis("off")

plt.tight_layout()
```

```python
from collections import Counter

labels = [label for _, label in dataset]
class_counts = Counter(labels)

for class_index, count in sorted(class_counts.items()):
    print(dataset.classes[class_index], count)
```
